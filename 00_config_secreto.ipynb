```python
from databricks.sdk import WorkspaceClient

w = WorkspaceClient()

try:
    w.secrets.create_scope(scope="kaggle")
    print("Scope 'kaggle' creado")
except Exception as e:
    print("Aviso al crear el scope:", e)

w.secrets.put_secret(scope="kaggle", key="api_token",
                     string_value=dbutils.widgets.get("token"))
print("Token guardado")

dbutils.widgets.remove("token")   # borra la cajita con el token
```
